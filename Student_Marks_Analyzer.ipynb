```python
import numpy as np

subjects = np.array([
    "Python", "NumPy", "Pandas",
    "Matplotlib", "Machine Learning"
])

marks = np.random.randint(35, 100, (50, 5))

print("Student Marks:")
print(marks)
print("Shape:", marks.shape)
print("Dimensions:", marks.ndim)
print("Total elements:", marks.size)
print("Data type:", marks.dtype)
# First 10 students
print("First 10 Students:")
print(marks[:10])

# Last subject marks for all students
print("Last Subject:")
print(marks[:, -1])

# Marks in the first 3 subjects for the first 10 students
print("First 10 Students, First 3 Subjects:")
print(marks[:10, :3])

# Marks of the last 5 students
print("Last 5 Students:")
print(marks[-5:, :])
# Marks below 40
fail_marks = marks[marks < 40]
print("Marks Below 40:")
print(fail_marks)

# Marks above 90
high_marks = marks[marks > 90]
print("Marks Above 90:")
print(high_marks)

# Grade bands
below_40 = np.sum(marks < 40)
grade_40_59 = np.sum((marks >= 40) & (marks < 60))
grade_60_79 = np.sum((marks >= 60) & (marks < 80))
grade_80_plus = np.sum(marks >= 80)

print("Below 40:", below_40)
print("40 to 59:", grade_40_59)
print("60 to 79:", grade_60_79)
print("80 and above:", grade_80_plus)
topper_indices = np.argmax(marks, axis=0)
topper_marks = np.max(marks, axis=0)

for i in range(len(subjects)):
    student_number = topper_indices[i] + 1

    print("Subject:", subjects[i])
    print("Topper: Student", student_number)
    print("Highest Marks:", topper_marks[i])
    print()
    pass_count = np.sum(marks >= 40, axis=0)
pass_percentage = (pass_count / 50) * 100

for i in range(len(subjects)):
    print(subjects[i])
    print("Students Passed:", pass_count[i])
    print("Pass Percentage:", pass_percentage[i], "%")
    print()
    print("===== STUDENT MARKS ANALYZER =====")
print("Total Students:", marks.shape[0])
print("Total Subjects:", marks.shape[1])
print("Total Marks Recorded:", marks.size)
print("Average Marks:", np.mean(marks))
print("Highest Mark:", np.max(marks))
print("Lowest Mark:", np.min(marks))
print("Overall Pass Percentage:",
      np.sum(marks >= 40) / marks.size * 100)
```
